```python
import numpy as np
import ModelNP
```

```python
data=np.load(r'dataset.npz',allow_pickle=True)
data_train,data_val=data['data_train'],data['data_val']
y_train,y_val=data['y_train'].astype(np.float32),data['y_val'].astype(np.float32)
y_train=np.expand_dims(y_train,axis=-1)
y_val=np.expand_dims(y_val,axis=-1)
model=ModelNP.LogisticModel()
BCE=ModelNP.BinariCrossEntropy()
optimizer=ModelNP.OptimizerSGD()
```

```python
ind=np.arange(len(data_train))
for epoh in range(30):
    np.random.shuffle(ind)
    data_train=data_train[ind]
    y_train=y_train[ind]
    for step in range(29):
        x=data_train[step:step+32]
        y_p=y_train[step:step+32]

        y=model(x)

        loss=BCE(y,y_p)
        
        grads=model.backward(y_p)
        grags=optimizer(grads)
        model.apply(grads)

    pred=model(data_val)
    accuracy=np.sum(((pred>0.5).astype(np.float32)==y_val))/y_val.shape[0]
    print(f'{epoh:2f} loss: {loss.tolist():.5f} accur: {accuracy.tolist():.4f}')
        
```
