```python
import numpy as np
import ModelNP
```

```python
data=np.load(r'dataset.npz',allow_pickle=True)
data_train,data_val=data['data_train'],data['data_val']
y_train,y_val=data['y_train'].astype(np.float32),data['y_val'].astype(np.float32)
y_train=np.expand_dims(y_train,axis=-1)
y_val=np.expand_dims(y_val,axis=-1)
model=ModelNP.LogisticModel()
BCE=ModelNP.BinariCrossEntropy()
optimizer=ModelNP.OptimizerSGD(0.005,0.7)
```

```python
ind=np.arange(len(data_train))
batch_size=64
accuracy_train=0
for epoh in range(3):
    np.random.shuffle(ind)
    data_train=data_train[ind]
    y_train=y_train[ind]
    for step in range(data_train.shape[0]//batch_size):
        x=data_train[step:step+batch_size]
        y_p=y_train[step:step+batch_size]

        y=model(x)

        loss=BCE(y,y_p)
        grads=model.backward(y_p)
        grags=optimizer(grads)
        model.apply(grads)
        accuracy_train+=np.sum(((y>0.5).astype(np.float32)==y_p))/y_p.shape[0]

        pred=model(data_val)
        accuracy=np.sum(((pred>0.5).astype(np.float32)==y_val))/y_val.shape[0]
        print(f'{epoh} step: {step} loss: {loss.tolist():.5f} accur: {accuracy.tolist():.4f} accuracy_train:{accuracy_train/(data_train.shape[0]//batch_size)}')
    accuracy_train=0
```
